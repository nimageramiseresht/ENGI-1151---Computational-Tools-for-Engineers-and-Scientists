#### 4.2.1 Engineering example: iterative calculation

````python
error = 1.0
tolerance = 0.01
iteration = 0

while error > tolerance:
    error = error / 2.0
    iteration += 1

    print("Iteration:", iteration, "Error:", error)

print("The tolerance has been satisfied.")
````

````python
error = 1.0
tolerance = 0.01
iteration = 0
maximum_iterations = 100

while error > tolerance and iteration < maximum_iterations:
    error = error / 2.0
    iteration += 1

if error <= tolerance:
    print("The calculation converged.")
else:
    print("The calculation did not converge.")
````

## 5 Controlling loops

### 5.1 The break statement

````python
for i in range(10):
    if i == 5:
        break

    print(i)
````

````python
temperatures = [22.0, 24.5, 27.1, 85.0, 25.3]
maximum_temperature = 80.0

for temperature in temperatures:
    if temperature > maximum_temperature:
        print("Unsafe temperature detected:", temperature)
        break

    print("Accepted temperature:", temperature)
````

### 5.2 The continue statement

````python
for i in range(5):
    if i == 2:
        continue

    print(i)
````

````python
measurements = [12.5, -999.0, 13.1, 12.8]
invalid_value = -999.0

for measurement in measurements:
    if measurement == invalid_value:
        continue

    print("Valid measurement:", measurement)
````

### 5.3 The pass statement

````python
temperature = 25.0

if temperature > 80.0:
    pass
else:
    print("The temperature is within the expected range.")
````

## 7 Exception handling

*This cell asks for input: type a value in the box that appears and press Enter.*

````python
try:
    number = int(input("Enter an integer: "))
    print("You entered:", number)

except ValueError:
    print("The entered value is not a valid integer.")
````

### 7.1 Handling more than one exception

*This cell asks for input: type a value in the box that appears and press Enter.*

````python
try:
    value = float(input("Enter a non-zero value: "))
    reciprocal = 1.0 / value

    print("Reciprocal:", reciprocal)

except ValueError:
    print("The input must be numerical.")

except ZeroDivisionError:
    print("The entered value must not be zero.")
````

### 7.2 The else clause

*This cell asks for input: type a value in the box that appears and press Enter.*

````python
try:
    value = float(input("Enter a numerical value: "))

except ValueError:
    print("The input must be numerical.")

else:
    print("The entered value is:", value)
````

### 7.3 The finally clause

*This cell asks for input: type a value in the box that appears and press Enter.*

````python
try:
    value = float(input("Enter a numerical value: "))
    print("The entered value is:", value)

except ValueError:
    print("The input must be numerical.")

finally:
    print("Input processing has finished.")
````
