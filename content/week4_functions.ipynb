## 4 Returning more than one value

```python
def basic_operations(a, b):
    addition = a + b
    subtraction = a - b
    multiplication = a * b

    if b == 0:
        division = None
    else:
        division = a / b

    return addition, subtraction, multiplication, division
```

```python
# Setup used by the next cells
def basic_operations(a, b):
    addition = a + b
    subtraction = a - b
    multiplication = a * b
    if b == 0:
        division = None
    else:
        division = a / b
    return addition, subtraction, multiplication, division
```

```python
results = basic_operations(3, 5)

print(results)
```

```python
addition, subtraction, multiplication, division = (
    basic_operations(3, 5)
)

print(addition)
print(subtraction)
print(multiplication)
print(division)
```

## 9 Nested functions

```python
def calculate_circle_properties(radius):

    def calculate_area():
        return np.pi * radius**2

    def calculate_circumference():
        return 2 * np.pi * radius

    area = calculate_area()
    circumference = calculate_circumference()

    return area, circumference
```

```python
# Setup used by the next cells
import numpy as np

def calculate_circle_properties(radius):

    def calculate_area():
        return np.pi * radius**2

    def calculate_circumference():
        return 2 * np.pi * radius

    area = calculate_area()
    circumference = calculate_circumference()

    return area, circumference
```

```python
area, circumference = calculate_circle_properties(5.0)

print("Area:", area)
print("Circumference:", circumference)
```
