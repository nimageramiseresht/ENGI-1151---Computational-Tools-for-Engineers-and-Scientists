### 7.4 Creating figures and axes explicitly

```python
import matplotlib.pyplot as plt
x = np.linspace(0, np.pi, 100)
fig, ax = plt.subplots(nrows=1, ncols=2)

ax[0].plot(x, np.sin(x))
ax[0].set_title("Sine Wave")
ax[0].set_xlabel("Angle, x [rad]")
ax[0].set_ylabel("sin(x)")
ax[0].grid(True)

ax[1].plot(x, np.cos(x))
ax[1].set_title("Cosine Wave")
ax[1].set_xlabel("Angle, x [rad]")
ax[1].set_ylabel("cos(x)")
ax[1].grid(False)
plt.show()
```

## 8 Creating three-dimensional surface plots

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

fig, ax = plt.subplots(
    subplot_kw={"projection": "3d"})

surface = ax.plot_surface(X, Y, Z, cmap=cm.viridis)
fig.colorbar(surface, shrink=0.6)

ax.set_title("Three-Dimensional Surface Plot")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")

plt.show()
```
